```python
from openai import OpenAI
from IPython.display import Audio, display
import tempfile

client = OpenAI()

texto = """
Último minuto.
Científicos de la Universidad de Ingeniería y Tecnología (UTEC) han presentado hoy un nuevo sistema de inteligencia artificial capaz de analizar grandes volúmenes de datos en solo segundos.
Según los investigadores, esta tecnología podría transformar sectores como la salud, la educación y la gestión pública.
Las pruebas iniciales han mostrado resultados prometedores y se espera que el proyecto entre en fase piloto durante los próximos meses.
"""

# 1. Generar el audio en WAV
response = client.audio.speech.create(
    model="gpt-4o-mini-tts",
    voice="verse",           # estilo reportero
    response_format="wav",
    input=texto
)

audio_bytes = response.read()

# 2. Guardar en un archivo temporal
with tempfile.NamedTemporaryFile(delete=False, suffix=".wav") as tmp:
    tmp.write(audio_bytes)
    tmp_path = tmp.name

# 3. Mostrar reproductor dentro del notebook
display(Audio(tmp_path))
```

```python
from openai import OpenAI
from IPython.display import Audio, display

import os
from dotenv import load_dotenv
import tempfile


# Cargar variables del archivo .env
load_dotenv()

client = OpenAI()


texto = """
Último minuto.

Científicos de la Universidad de Ingeniería y Tecnología (UTEC)
han presentado hoy un nuevo sistema de inteligencia artificial
capaz de analizar grandes volúmenes de datos en solo segundos.

Según los investigadores, esta tecnología podría transformar
sectores como la salud, la educación y la gestión pública.

Las pruebas iniciales han mostrado resultados prometedores
y se espera que el proyecto entre en fase piloto durante
los próximos meses.
"""


response = client.audio.speech.create(

    model="gpt-4o-mini-tts",

    voice="marin",

    input=texto,

    instructions="""
Habla en español latinoamericano neutro.

Usa el estilo de un presentador profesional de noticias de televisión.

Mantén un tono serio, claro, informativo y natural.

Habla con un ritmo ágil y fluido, ligeramente más rápido que una conversación normal.

Evita pausas largas entre frases. Mantén continuidad y energía en la narración.

Da especial énfasis a la frase "Último minuto", pero sin sonar exagerado.

Pronuncia con claridad, buena dicción y entonación periodística.

La locución debe sentirse dinámica, profesional y directa, como un boletín informativo de televisión.
""",

    response_format="wav",
)


audio_bytes = response.read()


with tempfile.NamedTemporaryFile(
    delete=False,
    suffix=".wav"
) as tmp:

    tmp.write(audio_bytes)

    tmp_path = tmp.name


display(
    Audio(tmp_path)
)
```
