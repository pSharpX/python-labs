```python
from openai import OpenAI
from dotenv import load_dotenv

import base64
import os


# ============================================================
# 1. CARGAR VARIABLES DE ENTORNO
# ============================================================

load_dotenv()

api_key = os.getenv("OPENAI_API_KEY")

if not api_key:
    raise ValueError(
        "No se encontró OPENAI_API_KEY en las variables de entorno."
    )


# ============================================================
# 2. CREAR CLIENTE DE OPENAI
# ============================================================

client = OpenAI(
    api_key=api_key
)


# ============================================================
# 3. PROMPT DE LA IMAGEN
# ============================================================

prompt = """
Fotografía periodística realista de un accidente de tránsito
entre dos automóviles en una avenida urbana.

Los dos vehículos presentan daños visibles después de una colisión.

Varios policías se encuentran en la escena ayudando a controlar
el tránsito y desviando los vehículos alrededor del accidente.

Debe observarse claramente:
- dos automóviles involucrados en el accidente,
- policías uniformados controlando el tránsito,
- conos de seguridad alrededor de la zona,
- otros vehículos circulando por la avenida,
- entorno urbano realista,
- iluminación natural de día.

Estilo fotográfico documental y periodístico.
Perspectiva amplia de la escena.
Sin heridas gráficas ni sangre.
"""


# ============================================================
# 4. GENERAR IMAGEN
# ============================================================

result = client.images.generate(
    model="gpt-image-1.5",

    prompt=prompt,

    n=1,

    size="1536x1024",

    quality="high",

    output_format="png",
)


# ============================================================
# 5. DECODIFICAR BASE64
# ============================================================

image_base64 = result.data[0].b64_json

image_bytes = base64.b64decode(
    image_base64
)


# ============================================================
# 6. GUARDAR IMAGEN
# ============================================================

output_path = "output1.png"

with open(output_path, "wb") as f:
    f.write(image_bytes)


print(
    f"Imagen generada correctamente: {output_path}"
)
```
